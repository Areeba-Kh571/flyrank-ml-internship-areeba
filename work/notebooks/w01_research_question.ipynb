## 2. The question: decision, action, cost of a wrong call

## The Question

**Research question:** Which published content pages should be reviewed first for refresh,
based on observable signals of staleness, declining visibility, and unmet traffic opportunity?

**Unit of analysis:** one row = one content page (a single published piece of content),
aggregated over a 90-day window.

**Output:** a ranked list of pages, ordered by a review-priority score, with reason codes
explaining why each page was flagged (e.g. "stale and visible," "declining with demand").

**The decision this improves:** a content reviewer with limited time needs to decide which
pages to look at first, out of potentially thousands.

**The action someone takes:** the reviewer opens the top-ranked pages and decides whether to
refresh, expand, or leave each one — the ranking itself doesn't take action, it directs
limited human attention.

**Cost of a wrong call:** if a page is ranked highly but wasn't actually a real opportunity,
the reviewer wastes limited time on it, and a genuinely declining page further down the list
may go unreviewed. This is a decision-support risk (wasted attention), not a safety risk.

**Why data/ML can help at all:** a hand-written rule can only combine a few conditions the
author already thought of. A model can weigh many observable signals (visibility, freshness,
position, CTR, engagement) simultaneously and rank pages by a learned combination — my own
notebook 01/02 runs already showed a random forest beating a hand rule (0.740 vs 0.240
Precision@50), suggesting real signal exists beyond simple hand-written conditions.
